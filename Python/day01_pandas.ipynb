# Day 01 — Pandas Fundamentals

This notebook documents the Pandas exercises completed for Day 1.

## 1. Dataset Creation

Create and display the employees DataFrame.

```python
import pandas as pd

employees = pd.DataFrame({
    "name": ["Ahmed", "Sara", "Omar", "Lina", "Yazan", "Noor", "Adam", "Dana"],
    "department": ["IT", "HR", "IT", "Sales", "IT", "HR", "Sales", "IT"],
    "salary": [700, 600, 900, 750, 1100, 650, 800, 950],
    "age": [24, 27, 29, 25, 31, 23, 28, 26]
})

employees
```

## 2. Basic Inspection

Display the first 5 rows using .head().

```python
employees.head()
```

## 3. Filtering

Find employees older than 25.

Find employees earning more than 800.

```python
older_than_25 = employees[employees["age"] > 25]
high_salary = employees[employees["salary"] > 800]

older_than_25

high_salary
```

## 4. Sorting

Sort employees by salary from highest to lowest.

```python
employees.sort_values(by="salary", ascending=False)
```

## 5. Aggregation

Calculate the overall average salary.

```python
employees["salary"].mean()
```

## 6. GroupBy

Calculate average salary by department.

Calculate maximum salary by department.

```python
employees.groupby("department")["salary"].mean()

employees.groupby("department")["salary"].max()
```

## 7. Creating a New Column

Create salary_level using .apply() and a lambda.

```python
employees["salary_level"] = employees["salary"].apply(lambda x: "High" if x >= 1000 else ("Medium" if x >= 700 else "Low"))

employees[["name", "salary", "salary_level"]]
```

## 8. Department Average

Calculate department average salary using:

groupby("department")["salary"].mean()

Map the department averages back onto the employees DataFrame as department_avg.

```python
dept_avg = employees.groupby("department")["salary"].mean()
employees["department_avg"] = employees["department"].map(dept_avg)

employees[["name", "department", "salary", "department_avg"]]
```

## 9. Above Department Average

Filter employees whose salary is greater than their department average.

```python
employees[employees["salary"] > employees["department_avg"]]
```

## 10. SQL → Pandas Translation

Include examples showing the Pandas equivalent of:

- WHERE filtering
- GROUP BY + AVG
- Above-department-average filtering

```python
# WHERE filtering
employees[employees["age"] > 25]

# GROUP BY + AVG
employees.groupby("department")["salary"].mean()

# Above-department-average filtering
employees[employees["salary"] > employees["department_avg"]]
```
